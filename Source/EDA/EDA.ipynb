# Xem outlier của các biến

```python
import numpy as np


outlier_cols = [
    'daily_vaccinations_per_million',
    'total_vaccinations_per_hundred',
    'people_fully_vaccinated_per_hundred',
    'vaccination_momentum'
]

log_df = df[outlier_cols].apply(
    lambda x: np.log1p(x)
)

plt.figure(figsize=(12, 6))

sns.boxplot(data=log_df,
            orient='h')

plt.xticks(rotation=25, ha='right')
plt.title('Boxplot các biến định lượng (Log1p)')
plt.ylabel('log(1 + value)')
plt.tight_layout()
plt.show()
```

# Top quốc gia theo % dân số đã tiêm

```python
latest = (
    df.sort_values(['country', 'date'])
    .groupby('country')
    .tail(1)
)
top10 = (
    latest[['country', 'people_vaccinated_per_hundred']]
    .sort_values('people_vaccinated_per_hundred', ascending=False)
    .head(10)
)
print("TOP 10:")
display(top10)
```

```python
plt.figure(figsize=(10,6))

ax = sns.barplot(
    data=top10,
    x='people_vaccinated_per_hundred',
    y='country'
)

plt.title('Top 10 quốc gia theo tỷ lệ dân số đã tiêm vaccine')
plt.xlabel('Tỷ lệ người đã tiêm chủng')
plt.ylabel('Quốc gia')

for container in ax.containers:
    ax.bar_label(container, fmt='%.1f%%',padding=3)
plt.tight_layout()
plt.show()
```

# Xem loại vaccine phổ biến

```python
vaccine_country = (
    df[['country', 'vaccine_names']]
    .drop_duplicates()
    .dropna()
)

vaccine_country = (
    vaccine_country
    .assign(vaccine_names=vaccine_country['vaccine_names'].str.split(','))
    .explode('vaccine_names')
)

vaccine_country['vaccine_names'] = (
    vaccine_country['vaccine_names'].str.strip()
)
vaccine_popularity = (
    vaccine_country
    .groupby('vaccine_names')['country']
    .nunique()
    .sort_values(ascending=False)
    .reset_index(name='country_count')
)

display(vaccine_popularity)
```

```python
plt.figure(figsize=(8, 5))

ax = sns.barplot(
    data=vaccine_popularity,
    x='country_count',
    y='vaccine_names'
)

plt.title('Các loại vaccine phổ biến')
plt.xlabel('Số quốc gia sử dụng')
plt.ylabel('Tên vaccine')
for container in ax.containers:
    ax.bar_label(container)
plt.tight_layout()
plt.show()
```

# Xem ma trận tương quan của các biến định lượng

```python
num_cols = [
    "total_vaccinations_per_hundred",
    "people_vaccinated_per_hundred",
    "people_fully_vaccinated_per_hundred",
    "daily_vaccinations_per_million",
    "dropout_date",
    "vaccination_momentum"
]
corr = df[num_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Ma trận tương quan")
plt.show()
```

# Xu hướng tiêm chủng theo thời gian

```python
global_daily = (
    df.groupby('date')['daily_vaccinations']
    .sum()
    .reset_index()
)

plt.figure(figsize=(12, 5))

sns.lineplot(
    data=global_daily,
    x='date',
    y='daily_vaccinations'
)

plt.title('Xu hướng tiêm chủng toàn cầu theo thời gian')
plt.xlabel('Ngày')
plt.ylabel('Tổng số liều tiêm mỗi ngày')

plt.tight_layout()
plt.show()
```

```python
momentum = (
    df.groupby('date')['vaccination_momentum']
    .mean()
    .reset_index()
)

plt.figure(figsize=(12, 5))

sns.lineplot(
    data=momentum,
    x='date',
    y='vaccination_momentum'
)

plt.axhline(0, linestyle='--')

plt.title('Xu hướng vaccination momentum theo thời gian')
plt.xlabel('Ngày')
plt.ylabel('Vaccination momentum')

plt.tight_layout()
plt.show()
```
