# Import thư viện

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../../Dataset/data_cleaned/country_vaccinations_added_fields.csv", index_col=0,
                 parse_dates=['date'])

df.head(10)
```

```python
df_check = df.sort_values(['country', 'date']).copy()
df_check['diff_total_vaccinations'] = (
    df_check.groupby('country')['total_vaccinations'].diff()
    )
print((df_check['diff_total_vaccinations'] < 0).sum())
df_check[df_check['diff_total_vaccinations']<0][
    ['country', 'date', 'total_vaccinations', 'diff_total_vaccinations']
]
```

# Xem outlier của các biến

```python
import numpy as np


outlier_cols = [
    'daily_vaccinations_per_million',
    'total_vaccinations_per_hundred',
    'people_fully_vaccinated_per_hundred',
    'vaccination_momentum'
]

log_df = df[outlier_cols].apply(
    lambda x: np.log1p(x)
)

plt.figure(figsize=(12, 6))

sns.boxplot(data=log_df,
            orient='h')

plt.xticks(rotation=25, ha='right')
plt.title('Boxplot các biến định lượng (Log1p)')
plt.ylabel('log(1 + value)')
plt.tight_layout()
plt.show()
```

# Xem phân bố các nhóm tỷ lệ bao phủ vaccine

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="coverage_group"
)

plt.title("Phân bố các nhóm tỷ lệ bao phủ vaccine")
plt.xlabel("Nhóm bao phủ vaccine")
plt.ylabel("Số lượng quan sát")
plt.show()
```

# Xem loại vaccine phổ biến

```python
vaccine_country = (
    df[['country', 'vaccine_names']]
    .drop_duplicates()
    .dropna()
)

vaccine_country = (
    vaccine_country
    .assign(vaccine_names=vaccine_country['vaccine_names'].str.split(','))
    .explode('vaccine_names')
)

vaccine_country['vaccine_names'] = (
    vaccine_country['vaccine_names'].str.strip()
)
vaccine_popularity = (
    vaccine_country
    .groupby('vaccine_names')['country']
    .nunique()
    .sort_values(ascending=False)
    .reset_index(name='country_count')
)

display(vaccine_popularity)
```

# Xem tỷ lệ bao phủ vaccine đầy đủ của các quốc gia/quan sát phân bố như thế nào.

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="people_fully_vaccinated_per_hundred",
    bins=30,
    kde=True
)

plt.title("Phân phối tỷ lệ dân số tiêm vaccine đầy đủ")
plt.xlabel("Người tiêm đầy đủ trên 100 dân (%)")
plt.ylabel("Số lượng quan sát")
plt.show()
```

# Xem ma trận tương quan của các biến định lượng

```python
num_cols = [
    "total_vaccinations_per_hundred",
    "people_vaccinated_per_hundred",
    "people_fully_vaccinated_per_hundred",
    "daily_vaccinations_per_million",
    "dropout_date",
    "vaccination_momentum"
]
corr = df[num_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Ma trận tương quan")
plt.show()
```

# Xu hướng tiêm chủng theo thời gian

```python
global_daily = (
    df.groupby('date')['daily_vaccinations']
    .sum()
    .reset_index()
)

plt.figure(figsize=(12, 5))

sns.lineplot(
    data=global_daily,
    x='date',
    y='daily_vaccinations'
)

plt.title('Xu hướng tiêm chủng toàn cầu theo thời gian')
plt.xlabel('Ngày')
plt.ylabel('Tổng số liều tiêm mỗi ngày')

plt.tight_layout()
plt.show()
```
