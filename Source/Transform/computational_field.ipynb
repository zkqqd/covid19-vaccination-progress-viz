## import dữ liệu

```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("country_vaccinations.csv")
df["date"]=pd.to_datetime(df["date"],errors="coerce")
df=df.sort_values(["country", "date"]).reset_index(drop=True)
df.head(10)
```

### 1. Drop-out rate
(people_vaccinated-people_fully_vaccinated) / people_vaccinated *100

```python
df["dropout_date"]=np.where(
    df["people_vaccinated"]>0,
    ((df["people_vaccinated"]-df["people_fully_vaccinated"])/df["people_vaccinated"])*100,
    np.nan
)
df.head(10)
```

## 2. Số lượng loại vaccine đang được sử dụng
Tách cột vaccines bằng dấu phẩy

```python
df["vaccine_count"]=df["vaccines"].apply(
    lambda x: len(
        [v.strip() for v in str(x).split(",") if v.strip()]
    )
)
df.head(10)
```

## 3. Phân nhóm tỉ lệ bao phủ vaccine
Thấp      : < 30%

Trung bình: 30% - 70%

Cao       : > 70%

```python
df["coverage_group"]=pd.cut(
    df["people_fully_vaccinated_per_hundred"],
    bins=[0, 30, 70, 100],
    labels=["Low", "Average", "High"],
    right=True,
    include_lowest=True
)
df.head(10)
```

## 4. Tốc độ tiêm chủng trung bình 7 ngày

```python
df["daily_vaccination_7d_avg"]=(
    df.groupby("country")["daily_vaccinations"]
      .transform(
          lambda x: x.rolling(
              window=7,
              min_periods=1
          ).mean()
      )
)
df.head(5)
```

5. Chỉ số đà tiêm chủng
So sánh số liều tiêm trong ngày với tổng số liều đã tiêm

```python
df["vaccination_momentum"] = (
    df["daily_vaccinations"]/ df["total_vaccinations"]* 100
)
df.head(10)
```
