## import dữ liệu

```python
import pandas as pd
import numpy as np
```

## Load dữ liệu

Ta sử dụng file đã merged để làm việc.

```python
df=pd.read_csv("../../Dataset/data_cleaned/country_vaccinations_merged.csv",  index_col=0)
df.head(10)
```

## Tạo các trường tính toán mới

### 1. Drop-out rate
(people_vaccinated-people_fully_vaccinated) / people_vaccinated *100

```python
df["dropout_date"]=np.where(
    df["people_vaccinated"]>0,
    ((df["people_vaccinated"]-df["people_fully_vaccinated"])/df["people_vaccinated"])*100,0
)
df.head(10)
```

### 2. Số lượng loại vaccine đang được sử dụng

```python
df["vaccine_count"] = df["vaccine_id"].apply(
    lambda x: len([v for v in str(x).split(",") if v.strip()])
)
df.head(10)
```

### 3. Phân nhóm tỉ lệ bao phủ vaccine
Thấp      : < 30%

Trung bình: 30% - 70%

Cao       : > 70%

```python
df["coverage_group"]=pd.cut(
    df["people_fully_vaccinated_per_hundred"],
    bins=[0, 30, 70, 100],
    labels=["Low", "Average", "High"],
    right=True,
    include_lowest=True
)
df.head(10)
```

### 4. Tốc độ tiêm chủng trung bình 7 ngày

```python
df["daily_vaccination_7d_avg"]=(
    df.groupby("country")["daily_vaccinations"]
      .transform(
          lambda x: x.rolling(
              window=7,
              min_periods=1
          ).mean()
      )
)
df.head(5)
```

### 5. Chỉ số đà tiêm chủng
So sánh số liều tiêm trong ngày với tổng số liều đã tiêm

```python
df["vaccination_momentum"] = df["daily_vaccinations"] / df["total_vaccinations"] * 100
df["vaccination_momentum"] = df["vaccination_momentum"].replace([np.inf, -np.inf], np.nan).fillna(0)
df.head(10)
```

## Lưu file dữ liệu mới

```python
df.to_csv("../../Dataset/data_cleaned/country_vaccinations_added_fields.csv")
print("\nĐã hoàn thành biến đổi dữ liệu!")
```
