## Import thư viện

```python
import pandas as pd
```

## Load dữ liệu

```python
splitted = pd.read_csv("../../Dataset/data_cleaned/country_vaccinations_cleaned_splitted.csv")
country = pd.read_csv("../../Dataset/data_cleaned/country.csv")
vaccine = pd.read_csv("../../Dataset/data_cleaned/vaccine.csv")
country_iso_vaccine = pd.read_csv("../../Dataset/data_cleaned/country_iso_vaccine.csv")
source = pd.read_csv("../../Dataset/data_cleaned/source.csv")
by_manufacturer = pd.read_csv("../../Dataset/data_cleaned/country_vaccinations_by_manufacturer_cleaned.csv")
```

## Join bảng splitted với country

```python
merged = splitted.merge(country, on="iso_code", how="left")
print(f"Sau khi join với country: {merged.shape}")
merged.head(3)
```

## Join bảng source

```python
merged = merged.merge(source, on="source_id", how="left")
print(f"Sau khi join với source: {merged.shape}")
merged.head(3)
```

## Join bảng country_iso_vaccine

```python
civ = (
    country_iso_vaccine[["iso_code", "vaccine_id"]]
    .drop_duplicates()
    .groupby("iso_code", as_index=False)["vaccine_id"]
    .agg(", ".join)
    .rename(columns={"vaccine_id": "vaccine_id"})
)

merged = merged.merge(civ, on="iso_code", how="left", validate="many_to_one")
print(f"Sau khi join với civ: {merged.shape}")
merged.head(3)
```

## Join bảng vaccine

```python
vaccine_map = dict(zip(vaccine["vaccine_id"], vaccine["vaccine_name"]))

merged["vaccine_names"] = merged["vaccine_id"].apply(
    lambda x: ", ".join(vaccine_map.get(i.strip(), "") for i in str(x).split(",") if i.strip())
)

print(f"Sau khi join với vaccine: {merged.shape}")
merged.head(3)
```

## Sắp xếp thứ tự cột 

```python
column_order = ['iso_code', 'country', 'date', 'total_vaccinations', 'people_vaccinated', 
                'people_fully_vaccinated', 'daily_vaccinations_raw', 'daily_vaccinations', 
                'total_vaccinations_per_hundred', 'people_vaccinated_per_hundred', 'people_fully_vaccinated_per_hundred',
                'daily_vaccinations_per_million', 'vaccine_id', 'vaccine_names', 
                'source_id', 'source_name', 'source_website']
merged = merged[column_order]
merged = merged.sort_values(["country", "date"]).reset_index(drop=True)
merged.head(3)
```

## Lưu file hoàn chỉnh

```python
merged.to_csv("../../Dataset/data_cleaned/country_vaccinations_merged.csv")
```
