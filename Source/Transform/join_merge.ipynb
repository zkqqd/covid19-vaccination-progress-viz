## Import thư viện

```python
import pandas as pd
```

## Load dữ liệu

```python
splitted = pd.read_csv("../../Dataset/data_cleaned/country_vaccinations_cleaned_splitted.csv")
country = pd.read_csv("../../Dataset/data_cleaned/country.csv")
vaccine = pd.read_csv("../../Dataset/data_cleaned/vaccine.csv")
country_iso_vaccine = pd.read_csv("../../Dataset/data_cleaned/country_iso_vaccine.csv")
source = pd.read_csv("../../Dataset/data_cleaned/source.csv")
by_manufacturer = pd.read_csv("../../Dataset/data_cleaned/country_vaccinations_by_manufacturer_cleaned.csv")
```

## Join bảng splitted với country

```python
merged = splitted.merge(country, on="iso_code", how="left")
print(f"Sau khi join với country: {merged.shape}")
merged.head(3)
```

## Join bảng source

```python
merged = merged.merge(source, on="source_id", how="left")
print(f"Sau khi join với source: {merged.shape}")
merged.head(3)
```

## Join bảng country_iso_vaccine

```python
civ = country_iso_vaccine[["iso_code", "vaccine_id"]].drop_duplicates()
merged = merged.merge(civ, on="iso_code", how="left")
print(f"Sau khi join với country_iso_vaccine: {merged.shape}")
merged.head(3)
```

## Join bảng vaccine

```python
merged = merged.merge(vaccine, on="vaccine_id", how="left")
print(f"Sau khi join với vaccine: {merged.shape}")
merged.head(3)
```

## Join bảng country_vaccinations_by_manufacturer_cleaned

```python
by_manu_renamed = by_manufacturer.rename(
    columns={
        "location": "country",
        "total_vaccinations": "manufacturer_total_vaccinations",
    }
)
merged = merged.merge(
    by_manu_renamed,
    on=["country", "date", "vaccine_name"],
    how="left",
)
print(f"Sau khi join với by_manufacturer: {merged.shape}")
merged.head(3)
```

## Sắp xếp thứ tự cột 

```python
column_order = [
    "iso_code", "country", "date",
    "total_vaccinations", "people_vaccinated", "people_fully_vaccinated",
    "daily_vaccinations_raw", "daily_vaccinations",
    "total_vaccinations_per_hundred", "people_vaccinated_per_hundred",
    "people_fully_vaccinated_per_hundred", "daily_vaccinations_per_million",
    "source_id", "source_name", "source_website",
    "vaccine_id", "vaccine_name", "manufacturer",
    "manufacturer_total_vaccinations",
]
merged = merged[column_order]
merged = merged.sort_values(["country", "date", "vaccine_name"]).reset_index(drop=True)
 
```
